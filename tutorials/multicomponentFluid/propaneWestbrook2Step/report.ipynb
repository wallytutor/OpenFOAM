# Introduction

In the present study we simulate propane-air combustion using the 2-step global mechanism proposed by [Westbrook (1981)](https://doi.org/10.1080/00102208108946970). The main goal of this case is providing a robust and fast convergence approach for establishing a steady-state solution in situations where simplified combustion is an acceptable simplification. Process conditions are provided in @tbl-summary-combustion and burner/domain characteristics in @tbl-system-dimensions.

```python
#| echo: false
#| tbl-cap: Summary of energy input conditions.
#| label: tbl-summary-combustion
supply = project.get_power_supply(1000)
_ = supply.report(notebook=True)
```

```python
#| echo: false
#| tbl-cap: System dimensions
#| label: tbl-system-dimensions
project.tabulate_dimensions()
```

Case is automated to support other operating conditions through a shared `initialConditions` file. Computed quantities related to boundary conditions for fuel and oxidizer are provided in @tbl-conditions-fuel and @tbl-conditions-oxid, respectively. Estimations of turbulent conditions assumed a turbulent intensity of 5%, while characteristic turbulent length scale is taken to be 7% of inlet equivalent diameter.

```python
#| echo: false
#| tbl-cap: Conditions evaluated for fuel.
#| label: tbl-conditions-fuel
project.flow_workflow(
    "fuel",
    supply.fuel_mass,
    supply.fuel_normal_density,
    298.15,
    project.AREA_FUEL,
    show=True
)
```

```python
#| echo: false
#| tbl-cap: Conditions evaluated for oxidizer.
#| label: tbl-conditions-oxid
project.flow_workflow(
    "oxid",
    supply.oxidizer_mass,
    supply.oxidizer_normal_density,
    298.15,
    project.AREA_OXIDIZER,
    show=True
)
```

An alternative scenario with pre-heated air is also simulated as provided in @ tbl-conditions-oxid-alt.

```python
#| echo: false
#| tbl-cap: Conditions evaluated for oxidizer.
#| label: tbl-conditions-oxid-alt
project.flow_workflow(
    "oxid",
    supply.oxidizer_mass,
    supply.oxidizer_normal_density,
    573.15,
    project.AREA_OXIDIZER,
    show=True
)
```

In all cases, walls are modeled through a global heat transfer coefficient representing a refractory of 280 mm (thermal conductivity 1.8 W/(m.K)) sourrounded by steady air at room temperature (heat transfer coefficient 10 W/(m².K)).
